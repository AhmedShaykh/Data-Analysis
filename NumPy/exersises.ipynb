```python
import numpy as np;
```

```python
sudo = np.array([ # Valid Sudoku
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]);

sudo
```

```python
int(sudo.sum()) # All Elements Sum
```

```python
np.sum(sudo, axis = 1) # 1 Axis Present Rows
```

```python
np.sum(sudo, axis = 0) # 0 Axis Present Columns
```

```python
checkRow = np.sum(sudo, axis = 1);

checkRow == 45
```

```python
checkCol = np.sum(sudo, axis = 0);

for i in checkCol:
    if i != 45:
        print("Sudoku Is Not Valid For Columns");
        break
else:
    print("Sudoku Is Valid For Columns");
```

```python
for i in checkRow:
    if i != 45:
        print("Sudoku Is Not Valid For Rows");
        break
else:
    print("Sudoku Is Valid For Rows");
```
