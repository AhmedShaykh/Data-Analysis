```python
import numpy as np;
```

# **Indexing & Slicing Of Vector**

```python
arr = np.arange(11, 21);

arr
```

```python
arr[2 : 6]
```

```python
arr_slice = arr[3 : : 2];

arr_slice
```

# **Indexing & Slicing Of Matrix**

```python
arr = np.arange(1,31).reshape(6,5);

arr
```

```python
arr[5] # Target On Row
```

```python
int(arr[0, 3]) # Target On Row & Column
```

```python
slice = arr[0 : 2, 1 : 3]

slice
```

```python
slice = arr[3 : , 2 : ]

slice
```

```python
arr[:, 2] # Target On Column
```

# **Boolean Indexing**

```python
arr = np.arange(11, 21);

arr
```

```python
bool_index = arr % 2 == 0; # Automatically Run Loop On Array

bool_index
```

```python
arr = arr[bool_index];

print(arr);
```
