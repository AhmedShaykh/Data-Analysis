# **NumPy Arithmetic Operations**

```python
a1 = np.array([1, 2, 3, 4, 5]);

a2 = np.array([6, 7, 8, 9, 10]);

print(a1);

print(a2);
```

```python
a1 + a2 # In Numpy Operators Can Run Direct Operations But Both Arrays Size Must Be Same
```

```python
a1 - a2 # No Need To Run Loop Function Apply Operators
```

```python
a1 * a2
```

```python
a1 / a2
```

```python
a1 // a2
```

```python
a1 ** a2
```

# **Broadcasting**

```python
l = [10, 20, 30, 40];

arr = np.array(l);

arr
```

```python
arr + 10
```

```python
arr2 = np.arange(1, 26).reshape(5, 5);

arr2
```

```python
arr2 = arr2 + 20;

arr2
```

```python
arr2 * 2
```

# **Deep & Shallow Copy**

```python
arr = np.arange(1, 21);

arr
```

```python
slice = arr[ : 5]; # Shallow Copy

slice
```

```python
slice = slice * 10; # Numpy Arrays Create Shallow Copy By Default

slice
```

```python
arr
```

```python
arr[4 : 6] * 10
```

```python
arr2 = arr;

arr2[0] = 99;

arr2
```

```python
arr  # Deep Copy
```

```python
arr = np.arange(1, 21);

arr

arr2 = arr.copy(); # Shallow Copy

arr2[0] = 99;

print(arr2);

print(arr);
```

# **Advance Array Manipulation**

```python
a = np.array([1, 2, 3, 4]); # Stacking Array

b = np.array([5, 6, 7, 8]); # You Can Create Stack Only Vector Array (Single) Not Matrix Array (Double)
```

```python
np.vstack((a, b)) # V Stack Is (Vertical Stack) Also Row Stack & In Stack Array Pass Must Be In Tuple
```

```python
np.column_stack((a,b)) # Convert Column Values Into Row Wise 2D Array
```

```python
np.hstack((a,b)) # Horizontal Stack (Row Wise)
```

```python
np.hsplit(arr, 4) # Horizontal Split
```
