```python
import pandas as pd;
```

# **Merging Dataframes**

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5],
    "name": ["John", "Anna", "Peter", "Linda", "Bob"],
    "department": ["HR", "IT", "Finance", "IT", "HR"]
});

salaries = pd.DataFrame({
    "employee_id": [1, 2, 3, 6, 7],
    "salary": [60000, 80000, 65000, 70000, 90000],
    "bonus": [5000, 10000, 7000, 8000, 12000]
});

employees
```

```python
salaries
```

```python
pd.merge(employees, salaries)
```

```python
pd.merge(employees, salaries, on = "employee_id") # "Employee_Id" Is The Common Column
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "inner") # Inner Is Default & Merges Both DataFrames Based On Common "Employee_Id" Column
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "outer") # Outer Join Keeps All "Employee_Id" Rows From Both DataFrames
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "left") # Left Join Keeps All Rows From The Employees DataFrame
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "right") # Right Join Keeps All Rows From The Salaries DataFrame
```

# **Concatination Dataframes**

```python
df1 = pd.DataFrame({
    "A": ["A0", "A1", "A2"],
    "B": ["B0", "B1", "B2"],
    "C": ["C0", "C1", "C2"]
});

df2 = pd.DataFrame({
    "A": ["A3", "A4", "A5"],
    "B": ["B3", "B4", "B5"],
    "C": ["C3", "C4", "C5"]
});

df1
```

```python
df2
```

```python
pd.concat([df2, df1]) # Vertical (Column Wise)
```

```python
pd.concat([df2, df1], axis = 1) # Horizontal (Row Wise)
```

# **Joining Dataframes**

```python
df1 = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[1, 2, 3]);

df2 = pd.DataFrame({
    "score": [85, 90, 75]
}, index=[2, 3, 4]);

df1
```

```python
df2
```

```python
df1.join(df2) # Default Inner Join
```

```python
df1.join(df2, how = "outer")
```

```python
df2.join(df1)
```
