```python
import pandas as pd;
import numpy as np;
```

# **Pandas Series**

```python
labels = ["a", "b", "c"];
          
li = [10, 20, 30];

arr = np.array([10, 20, 30]);

dic = { "a" : 10, "b" : 20, "c" : 30 };
```

```python
pd.Series(li) # Series Creates A Single Column With Automatic Index Label
```

```python
pd.Series(li, index=labels) # Can Set With Custom Labels Also
```

```python
pd.Series(arr, index=labels)
```

```python
pd.Series(dic) # Creates Series With Using Dictionary Keys As Index labels
```

# **Pandas DataFrames**

```python
data = {
    "Name": ["Ahmed", "Asmir", "Jad", "Emmett"],
    "Age": [25, 39, 38, 25],
    "City": ["Karachi", "Boras", "Gothenburg", "Amsterdam"],
    "Salary": [90000, 80000, 72000, 125000]
};

df = pd.DataFrame(data);

df
```

# **Missing Data**

```python
data = {
    "A": [1, 2, np.nan, 4, 5],
    "B": [1, 2, 3, 4, 5],
    "C": [1, 2, 3, np.nan, np.nan],
    "D": [1, np.nan, np.nan, np.nan, 5]
};

df3 = pd.DataFrame(data);
```

```python
df3
```

```python
df3.isna() # Finding Missing Data (Null Values)
```

```python
df3.isna().sum()
```

```python
df3.isna().any() # Check Column All Values Is Null
```

```python
df3
```

```python
df3.dropna() # Remove Null Values In Row
```

```python
df3
```

```python
df3.dropna(thresh = 3) # Atleast 3 Non-Null Values
```

```python
df3.dropna(thresh = 1) # Atleast 1 Non-Null Value
```

```python
df3
```

```python
df3.fillna(0) # Filling Missing Data In Null Values
```

```python
df3
```

```python
values = { "A" : 0, "B" : 100, "C" : 300, "D" : 400 };

df3.fillna(value = values) # It Work In Columns
```

```python
df3
```

```python
df3.fillna(df3.mean()) # Fill Missing Values With Column Mean.
```
